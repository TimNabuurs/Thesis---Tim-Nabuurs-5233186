```python
import pandas as pd
import numpy as np

# --------------------------------------------------------------------------EXTERNAL DATA SECTION----------------------------------------------------------------------------------------- # 

# --- 0. DATA PREPARATION ---
file_path = 'Input Temp & Rad - from dynamic excel model.csv'
hourly_df = pd.read_csv(file_path, sep=';')

# --------------------------------------------------------------------------SCHEDULE SECTION----------------------------------------------------------------------------------------- # 

# --- PROGRAMMATIC SCHEDULES (168-hour design week profiles) - values from Excel "Schedules" tab---
# 1. Occupancy (Presence of people) 
occ_profile = [
    # Monday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Tuesday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Wednesday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Thursday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Friday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Saturday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Sunday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50
]

# 2. Lighting & Appliances
equip_profile = [
    # Monday
    0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Tuesday
    0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Wednesday
    0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Thursday
    0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Friday
    0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Saturday
    0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Sunday
    0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.10, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50
]

# 3. Ventilation Airflow Schedule (mechanical system operation)
vent_profile = [
    # Monday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Tuesday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Wednesday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Thursday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Friday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.20, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Saturday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50,
    # Sunday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50
]

# 4. Heating/Cooling On/Off Schedule (system state - operating hours)
system_profile = [
    # Monday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00,
    # Tuesday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00,
    # Wednesday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00,
    # Thursday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00,
    # Friday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00,
    # Saturday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00,
    # Sunday
    0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 0.50, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00
]

# --------------------------------------------------------------------------INPUT BUILDING GEOMETRY----------------------------------------------------------------------------------------- # 

# --- 1. SETTINGS & BUILDING PROPERTIES ---
# Geometry zone
floor_area = 3690.0                           # m2
room_height = 3.0                             # m
room_volume = floor_area * room_height        # m3
area_ground = 410.0                           # m2

# Building Surface Areas 
total_facade_areas = {
    'N': 0.0, 'NE': 836.00, 'E': 0.0, 'SE': 390.00,
    'S': 0.0, 'SW': 836.00, 'W': 0.0, 'NW': 390.00
} # m2
window_percentages = {
    'N': 0.00, 'NE': 0.27, 'E': 0.00, 'SE': 0.33,
    'S': 0.00, 'SW': 0.36, 'W': 0.00, 'NW': 0.33
} # [%]
glazing_percentages = {
    'N': 0.00, 'NE': 0.80, 'E': 0.00, 'SE': 0.80,
    'S': 0.00, 'SW': 0.80, 'W': 0.00, 'NW': 0.80
} # [%]

# Calculate derived areas 
window_areas = {k: (total_facade_areas[k] * window_percentages[k]) for k in total_facade_areas}
glass_areas = {k: (total_facade_areas[k] * window_percentages[k] * glazing_percentages[k]) for k in total_facade_areas}
opaque_wall_areas = {k: (total_facade_areas[k] - window_areas[k]) for k in total_facade_areas}

area_windows = sum(window_areas.values())    # m2
area_walls = sum(opaque_wall_areas.values()) # m2
area_roof  = 410.0                           # m2
area_floor = 410.0                           # m2

# Transmission Properties (For Rc to U-Value Conversion)
rc_facade = 3.9              # m2K/W
rc_roof = 5.7                # m2K/W
rc_ground_floor = 3.0        # m2K/W
u_value_windows = 1.8        # W/m2K

# --------------------------------------------------------------------------INPUT SOLAR----------------------------------------------------------------------------------------- # 

# Surface coefficients and solar absorption
alfai = 7.5                                 # W/m2K (convection/radiation coefficient indoor)
alfao = 15.0                                # W/m2K (convection/radiation coefficient outdoor)
solar_absorption_coefficient = 0.5          # - (absorption coefficient)
solar_heat_coefficient_shading = 0.5        # - (shading/reduction factor for solar radiation)
solar_heat_coefficient_glazing = 0.6        # - (G-value)

# Orientations matching headers from 'Separate input datasets from dynamic model excel.csv' exactly!
solar_orientations = ['N', 'NE', 'E', 'SE', 'S', 'SW', 'W', 'NW']

# --------------------------------------------------------------------------INPUT VENTILATION----------------------------------------------------------------------------------------- # 

# Ventilation & Infiltration Properties
air_density = 1.2              # kg/m3
air_heat_capacity = 1003.0     # J/kgK
vent_flow_per_person = 36.0    # m3/h/pp
infiltration_ach = 0.2         # h-1
natural_vent_rate = 3.0        # h-1
vent_cooling_setpoint = 23.9   # oC
eta = 0                        # - (Heat recovery efficiency for ventilation)

# --------------------------------------------------------------------------INPUT INTERNAL GAINS AND FLOW RATES---------------------------------------------------------------------------------------- # 

# Internal Heat Gains Constants
max_people_per_m2 = 0.02                        # pp/m2
max_people = max_people_per_m2 * floor_area     # pp
heat_per_person = 65.0                          # W/pp
appliances_w_m2 = 6.0                           # W/m2
lighting_w_m2 = 6.0                             # W/m2

# Ventilation flow rate based on occupancy (Coupling)
flow_rate_vent = max_people * vent_flow_per_person                              # m3/h 
total_flow_rate_vent = vent_flow_per_person * max_people_per_m2 * floor_area    # m3/h 

# Ventilation system properties
system_pressure_drop = 200.0               # Pa
heat_recovery_efficiency = 0.0             # - (0 for no heat recovery, 1 for perfect heat recovery)
efficiency_fan_and_motor = 0.6             # - (Efficiency of the fan and motor system)

fans_power = total_flow_rate_vent * system_pressure_drop / (3600 * efficiency_fan_and_motor)

# --------------------------------------------------------------------------INPUT CONSTANTS----------------------------------------------------------------------------------------- # 

# Simulation Constants
thermal_capacity = 1000 * floor_area  # J/K (Matches 1 kJ/m2K from Excel)
heating_setpoint = 20.0               # oC
cooling_setpoint = 24.0               # oC
dt = 3600                             # s
current_temp = 20.0                   # oC
heating_power_max = 1000000.0         # W (Positive for heating)
cooling_power_max = -1000000.0        # W (Negative for cooling)
t_ground = 15.0                       # Ground sink temperature

# --------------------------------------------------------------------------U-values based on NTA8800----------------------------------------------------------------------------------------- # 

# U-value calculation due to thermal bridges based on NTA8800 curve
def calculate_nta8800_u_value(rc_value):
    if rc_value <= 0:
        return 0
        
    # 1. Calculate undisturbed U-value 
    u_undisturbed = 1 / (rc_value + 0.17)
    
    # 2. Calculate the Delta U penalty using the exact Excel formula
    delta_u = 0.15                   # Either use delta_u = max(0, 0.1 - 0.25 * (u_undisturbed - 0.4)) for accuracy, or the 0.15 fixed penalty as is done in the excel.
    
    # 3. Final U-value
    return u_undisturbed + delta_u

#Accurate U-values for walls, roof, and floor based on NTA8800 curve
u_value_walls = calculate_nta8800_u_value(rc_facade)        
u_value_roof  = calculate_nta8800_u_value(rc_roof) 
u_value_ground = calculate_nta8800_u_value(rc_ground_floor)       

# --- 2. DYNAMIC PHYSICS CALCULATION (Couplings) ---
# Transmission coupling
trans_e_windows = area_windows * u_value_windows
trans_e_walls = area_walls * u_value_walls
trans_e_roof = area_roof * u_value_roof

# Total transmission with external outdoor temperature
H_e_transmission = trans_e_walls + trans_e_roof + trans_e_windows # W/K

# --------------------------------------------------------------------------CALCULATION LOOP SECTION----------------------------------------------------------------------------------------- # 

# --- 3. THE 8760 HOURLY LOOP ---
annual_heating_results = []
annual_cooling_results = []
annual_temp_results = []
debug_data = [] # List to store hourly rows for export

for index, row in hourly_df.iterrows():
    # --- DESIGN WEEK LOGIC: Pull values from 168-hour profiles ---
    idx_168 = index % 168
    current_occ_val = occ_profile[idx_168]
    current_light_val = equip_profile[idx_168]
    current_vent_val = vent_profile[idx_168]
    current_sys = system_profile[idx_168]

    # --------------------------------------------------------------------------LOOP: SOLAR GAINS----------------------------------------------------------------------------------------- # 

    # Dictionary
    glazing_gains_breakdown = {}
    opaque_gains_breakdown = {}

    for orient in solar_orientations:
        # --- DYNAMIC SHADING ALTERATION START ---
        # Logic: IF(AND(T_i_prev > 23.9 ; Radiation > 0) ; 0.5 ; 1.0)
        if current_temp >= 23.9 and row[orient] > 0:
            active_shading = solar_heat_coefficient_shading # This is the 0.5 factor
        else:
            active_shading = 1.0 # No shading
        # --- DYNAMIC SHADING ALTERATION END ---

        # Glazing Gain per orientation (W) 
        glazing_val = (row[orient] * glass_areas[orient] * solar_heat_coefficient_glazing * active_shading)
        glazing_gains_breakdown[orient] = glazing_val

        # Opaque Gain per orientation (W)
        opaque_val = (row[orient] * opaque_wall_areas[orient] * solar_absorption_coefficient * (u_value_walls / alfao))
        opaque_gains_breakdown[orient] = opaque_val

    # Total sums 
    sun_glazing = sum(glazing_gains_breakdown.values())
    sun_opaque_walls = sum(opaque_gains_breakdown.values())
    sun_opaque_roof = row['Horizontal'] * solar_absorption_coefficient * (u_value_roof / alfao) * area_roof

    # --------------------------------------------------------------------------LOOP: INTERNAL GAINS----------------------------------------------------------------------------------------- # 

    people_heat = max_people * current_occ_val * heat_per_person
    lighting_heat = lighting_w_m2 * floor_area * current_light_val
    equipment_heat = appliances_w_m2 * floor_area * current_light_val
    fans_heat = 0.5 * fans_power * current_vent_val                                             # Assuming 50% of fan power converts to heat in the space 

    # Total sum
    total_internal_gains = people_heat + equipment_heat + lighting_heat + fans_heat * 0.5       # The excel adds another 0.5 factor (not sure why?)

    # --------------------------------------------------------------------------LOOP: TOTAL FREE GAINS----------------------------------------------------------------------------------------- # 

    total_free_gains = sun_glazing + sun_opaque_walls + sun_opaque_roof + total_internal_gains

    # --------------------------------------------------------------------------LOOP: GROUND AND INFILTRATION----------------------------------------------------------------------------------------- # 

    # Flow rate infiltration
    flow_rate_inf = room_volume * infiltration_ach     # m3/h

    # Infiltration and Ground Coupling
    h_inf = (flow_rate_inf * air_density * air_heat_capacity) / 3600
    h_ground = area_ground * u_value_ground

    # --------------------------------------------------------------------------LOOP: HEAT TRANSFER VENTILATION----------------------------------------------------------------------------------------- # 

    current_flow_m3h = flow_rate_vent * current_occ_val * current_vent_val     

    # Ventilation coupling
    h_vent_heat = (1 - eta) * (flow_rate_vent * current_vent_val * air_density * air_heat_capacity) / 3600
    h_vent_cool = ((flow_rate_vent * current_vent_val) + (natural_vent_rate * room_volume)) * air_density * air_heat_capacity / 3600

    # --------------------------------------------------------------------------LOOP: TOTAL HEAT TRANSFER----------------------------------------------------------------------------------------- # 
    # External (outdoor) temperature
    t_ext = row['T'] 

    # Total loss with external temperature 
    H_total_heat = H_e_transmission + h_vent_heat + h_inf        
    H_total_cool = H_e_transmission + h_vent_cool + h_inf          

    # Total loss with external temperature & ground temperature (Indoor temperature not included as it is the variable we are trying to predict)
    Total_exterior_heat = (H_total_heat * t_ext) + (h_ground * t_ground)
    Total_exterior_cool = (H_total_cool * t_ext) + (h_ground * t_ground)

    # Total Coupling for the denominator (Now MUST include h_ground)
    H_total_with_ground_heat = H_total_heat + h_ground
    H_total_with_ground_cool = H_total_cool + h_ground
    
    # --------------------------------------------------------------------------LOOP: DYNAMIC VENTILATION LOGIC----------------------------------------------------------------------------------------- # 
    # The Prediction
    den_test = 1 + (dt / thermal_capacity) * H_total_with_ground_heat
    
    # T_check: Predicted indoor temperature based on current state
    t_check = (current_temp + (dt / thermal_capacity) * (total_free_gains + Total_exterior_heat)) / den_test

    # T_eq: The equilibrium "Target" temperature as defined in the PDF background info
    # Equation: (Internal Gains + Driving Gains) / Total Couplings
    t_eq_heat = (total_free_gains + Total_exterior_heat) / H_total_with_ground_heat
    t_eq_cool = (total_free_gains + Total_exterior_cool) / H_total_with_ground_cool

    # The Decision (Night Cooling / Bypass Logic)
    # Is it too hot AND cooler outside? 
    if t_check > vent_cooling_setpoint and t_ext < current_temp:
        H_active = H_total_with_ground_cool  # Use the high-ventilation coupling + ground
        current_ach = (current_flow_m3h / room_volume) + natural_vent_rate
    else:
        H_active = H_total_with_ground_heat  # Use the low-ventilation/WTW coupling + ground
        current_ach = (current_flow_m3h / room_volume)

    # --------------------------------------------------------------------------LOOP: EXPONENTIALS----------------------------------------------------------------------------------------- # 

    # Calculate Exponential Factors for Heat and Cool modes
    exp_factor_heat = 1 - np.exp(-(H_total_with_ground_heat) / thermal_capacity * dt)        
    exp_factor_cool = 1 - np.exp(-(H_total_with_ground_cool) / thermal_capacity * dt)        

    # --------------------------------------------------------------------------LOOP: SCOUTING TEMPERATURES ----------------------------------------------------------------------------------------- # 

    # T_room_free: Temperature reached with 0 W heating/cooling
    t_room_free_heating = current_temp + (t_eq_heat - current_temp) * exp_factor_heat
    t_room_free_cooling = current_temp + (t_eq_cool - current_temp) * exp_factor_cool

    # T_max: Temperature reached if system runs at max capacity
    # (Power / H) gives the degrees added; the exp_factor handles the time delay
    t_room_max_heating = t_room_free_heating + (heating_power_max / H_total_with_ground_heat) * exp_factor_heat
    t_room_max_cooling = t_room_free_cooling + (cooling_power_max / H_total_with_ground_cool) * exp_factor_cool     # cooling_power_max is already negative, therefore the '+' sign

    # --------------------------------------------------------------------------LOOP: INITIALIZE----------------------------------------------------------------------------------------- #

    # Initialize results
    q_heat_kwh = 0.0
    q_cool_kwh = 0.0

    # --------------------------------------------------------------------------LOOP: CONTROL LOGIC GATE----------------------------------------------------------------------------------------- #

    # 5. D. THE LOGIC GATE
    system_active = current_sys

    if system_active > 0:
        # STEP 1: Heating Capacity Limit (Too cold even with max heat)
        # If T(t+Δt)heat < Tset,h
        if t_room_max_heating < heating_setpoint:
            current_temp = t_room_max_heating
            q_heat_kwh = heating_power_max / 1000.0
            q_cool_kwh = 0.0
            
        # STEP 2: Heating Setpoint Reached (Modulating power)
        # If T(t+Δt)heat > Tset,h and T(t+Δt)free,heat < Tset,h
        elif t_room_max_heating >= heating_setpoint and t_room_free_heating < heating_setpoint:
            current_temp = heating_setpoint
            # Simplified ISO formula: Power needed to close the gap from 'free' to 'setpoint'
            p_needed = H_total_with_ground_heat * (heating_setpoint - t_room_free_heating) / exp_factor_heat
            q_heat_kwh = p_needed / 1000.0
            q_cool_kwh = 0.0
            
        # STEP 3: Free Floating - Standard Ventilation (HVAC is idle)
        # If T(t+Δt)free,heat > Tset,h and T(t+Δt)free,heat < Tset,v
        elif t_room_free_heating >= heating_setpoint and t_room_free_heating <= vent_cooling_setpoint:
            current_temp = t_room_free_heating
            q_heat_kwh = 0.0
            q_cool_kwh = 0.0
            
        # STEP 4: Ventilation Setpoint Reached (Windows/Bypass modulating to hold 23.9)
        # If T(t+Δt)free,heat > Tset,v and T(t+Δt)free,cool < Tset,v
        elif t_room_free_heating > vent_cooling_setpoint and t_room_free_cooling < vent_cooling_setpoint:
            current_temp = vent_cooling_setpoint
            q_heat_kwh = 0.0
            q_cool_kwh = 0.0
            
        # STEP 5: Free Floating - High Ventilation (Night cooling active but not enough)
        # If T(t+Δt)free,cool > Tset,v and T(t+Δt)free,cool < Tset,c
        elif t_room_free_cooling >= vent_cooling_setpoint and t_room_free_cooling <= cooling_setpoint:
            current_temp = t_room_free_cooling
            q_heat_kwh = 0.0
            q_cool_kwh = 0.0
            
        # STEP 6: Cooling Setpoint Reached (Modulating chiller)
        # If T(t+Δt)cool < Tset,c and T(t+Δt)free,cool > Tset,c
        elif t_room_max_cooling <= cooling_setpoint and t_room_free_cooling > cooling_setpoint:
            current_temp = cooling_setpoint
            
            # Logic Check: Use H_total_with_ground_cool ONLY if natural ventilation is physically possible/beneficial.
            # If the chiller is working against outside air that is HOTTER than the room, 
            # the Excel tool defaults back to the standard coupling to avoid "cooling the outdoors."
            
            if t_ext < cooling_setpoint:
                # Natural vent helps; use the higher coupling
                p_needed = H_total_with_ground_cool * (cooling_setpoint - t_room_free_cooling) / exp_factor_cool
            else:
                # Natural vent hurts; Excel logic typically assumes windows closed during active chilling
                p_needed = H_total_with_ground_heat * (cooling_setpoint - t_room_free_heating) / exp_factor_heat
                
            q_cool_kwh = p_needed / 1000.0
            q_heat_kwh = 0.0
            
        # STEP 7: Cooling Capacity Limit (Too hot even with max cooling)
        # If T(t+Δt)cool > Tset,c
        elif t_room_max_cooling > cooling_setpoint:
            current_temp = t_room_max_cooling
            q_cool_kwh = cooling_power_max / 1000.0
            q_heat_kwh = 0.0
            
    else:
        # System is OFF (e.g., Night/Weekend schedule)
        # Building floats based on the most recent physics state (heating-mode coupling)
        current_temp = t_room_free_heating
        q_heat_kwh = 0.0
        q_cool_kwh = 0.0

    # --------------------------------------------------------------------------LOOP: STORE RESULTS----------------------------------------------------------------------------------------- #

    annual_heating_results.append(q_heat_kwh)
    annual_cooling_results.append(q_cool_kwh)
    annual_temp_results.append(current_temp)

    # --------------------------------------------------------------------------OUTPUT SECTION----------------------------------------------------------------------------------------- #

    # --- NEW: Store values for export ---
    row_data = {    
        'Hour': index + 1,
        'T_ext': t_ext,

        # 1. Individual Couplings
        'Trans window': trans_e_windows,
        'Trans wall': trans_e_walls,
        'Trans roof': trans_e_roof,
        'Transmission tot (W/K)': H_e_transmission,
        'Infiltration (W/K)': h_inf,
        'Ventilation heat mode (W/K)': h_vent_heat,
        'Ventilation cool mode (W/K)': h_vent_cool,
        
        # 2. Total Couplings (Excel Columns Q & R)
        'Heating mode tot coupling (W/K)': H_total_heat,
        'Cooling mode tot coupling (W/K)': H_total_cool,
        'Ground coupling (W/K)': h_ground,
        
        # 3. Solar and Internal Gains
        'Solar glazing (Wh)': sun_glazing,
        'Solar opaque (Wh)': sun_opaque_walls + sun_opaque_roof,
        'People (Wh)': people_heat,
        'Lighting (Wh)': lighting_heat,
        'Appliances (Wh)': equipment_heat,
        'Fans (Wh)': fans_heat,
        'Internal heat gains (Wh)': total_internal_gains,

        # 4. Temperature "Scouts" (Excel Columns AJ, AK, AL, AM)
        'T_i Max heating': t_room_max_heating,
        'T_i free floating with H_heat': t_room_free_heating,
        'T_i Max cooling': t_room_max_cooling,
        'T_i free floating with H_cool': t_room_free_cooling,
        'T_i indoor (oC)': current_temp,
        
        # 5. Final Results (Excel Columns AN, AO, AP)
        'Q_heat (kWh)': q_heat_kwh,
        'Q_cool (kWh)': q_cool_kwh,

        # 6. Schedules (Last)
        'Sch_Occupancy': current_occ_val,
        'Sch_Equipment': current_light_val,
        'Sch_SystemActive': current_sys,
        'Sch_Ventilation': current_vent_val
    }

    # --- ADD PER-ORIENTATION RADIATION DATA iN OUTPUT FILE ---
    for orient in solar_orientations:
        row_data[f'G_{orient}_(Wh)'] = glazing_gains_breakdown[orient]

    for orient in solar_orientations:
        row_data[f'O_{orient}_(Wh)'] = opaque_gains_breakdown[orient]
    
    row_data['Opaque_Roof_(W)'] = sun_opaque_roof

    debug_data.append(row_data)

# --------------------------------------------------------------------------EXPORT SECTION----------------------------------------------------------------------------------------- # 

# --- 4. OUTPUTS ---
print(f"Simulation Complete.")
print(f"Room Volume: {room_volume:.2f} m3")
print(f"Peak Heating: {max(annual_heating_results):.2f} kW | Peak Cooling: {min(annual_cooling_results):.2f} kW")
print(f"Total Annual Heat Demand: {sum(annual_heating_results):.2f} kWh")
print(f"Total Annual Cool Demand: {sum(annual_cooling_results):.2f} kWh")
print(f"Roof U-Value: {u_value_roof:.4f} W/m2/K")
print(f"Facade wall U-Value: {u_value_walls:.4f} W/m2/K")
print(f"Ground U-Value: {u_value_ground:.4f} W/m2/K")
print(f"Check fans: {fans_power:.2f} kW")


# Create DataFrame and Export to CSV
export_df = pd.DataFrame(debug_data)

# Round all numerical values to 2 decimal places for readability
export_df = export_df.round(2)

# Define the filename
excel_filename = 'Dynamic_Model_Comparison.xlsx'

# Export to Excel with formatting
with pd.ExcelWriter(excel_filename, engine='openpyxl') as writer:
    export_df.to_excel(writer, index=False, sheet_name='Python_Results')
    
    # Access the openpyxl worksheet object to adjust column widths
    worksheet = writer.sheets['Python_Results']
    for column_cells in worksheet.columns:
        # Determine length based on cell content
        content_length = max(len(str(cell.value)) for cell in column_cells)
        
        # Column width
        adjusted_width = min(max(content_length, 10), 25)
        
        worksheet.column_dimensions[column_cells[0].column_letter].width = adjusted_width
```
